```python
df_clean = df.drop(columns=["Year", "Surname"])
```

```python
df_clean.columns
```

```python
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[0, 29, 45, 60, 120],
    labels=["<30", "30-45", "46-60", "60+"]
)
```

```python
df_clean["AgeGroup"].value_counts().sort_index()
```

```python
df_clean.groupby("AgeGroup", observed=True)["Exited"].mean() * 100
```

```python
df_clean["CreditBand"] = pd.cut(
    df_clean["CreditScore"],
    bins=[0, 579, 669, 900],
    labels=["Low", "Medium", "High"]
)
```

```python
df_clean["TenureGroup"] = pd.cut(
    df_clean["Tenure"],
    bins=[-1, 2, 6, 10],
    labels=["New", "Mid-term", "Long-term"]
)
```

```python
print(df_clean["CreditBand"].value_counts().sort_index())
print(df_clean["TenureGroup"].value_counts().sort_index())
```

```python
print(df_clean.groupby("CreditBand", observed=True)["Exited"].mean() * 100)
print(df_clean.groupby("TenureGroup", observed=True)["Exited"].mean() * 100)
```

```python
nonzero_median = df_clean.loc[df_clean["Balance"] > 0, "Balance"].median()

def balance_segment(b):
    if b == 0:
        return "Zero-balance"
    elif b <= nonzero_median:
        return "Low-balance"
    else:
        return "High-balance"

df_clean["BalanceSegment"] = df_clean["Balance"].apply(balance_segment)
```

```python
print(df_clean["BalanceSegment"].value_counts())
print(df_clean.groupby("BalanceSegment")["Exited"].mean() * 100)
```

```python
print(df_clean.groupby("Geography")["Exited"].mean() * 100)
print(df_clean.groupby("Gender")["Exited"].mean() * 100)
```

```python
geo = df_clean.groupby("Geography").agg(
    customers=("Exited", "count"),
    churners=("Exited", "sum"),
    churn_rate=("Exited", "mean")
)

geo["churn_rate"] = geo["churn_rate"] * 100
geo["share_of_customers"] = geo["customers"] / geo["customers"].sum() * 100
geo["share_of_churn"] = geo["churners"] / geo["churners"].sum() * 100
geo["risk_index"] = geo["share_of_churn"] / geo["share_of_customers"]

geo
```

```python
df_clean.groupby("Exited")[["Age", "Balance", "EstimatedSalary", "CreditScore", "NumOfProducts", "Tenure"]].mean()
```

```python
pd.pivot_table(
    df_clean,
    values="Exited",
    index="Geography",
    columns="AgeGroup",
    aggfunc="mean",
    observed=True
) * 100
```

```python
threshold = df_clean["Balance"].quantile(0.75)
df_clean["HighValue"] = df_clean["Balance"] >= threshold
threshold
```

```python
hv = df_clean[df_clean["HighValue"]]
print("High-value customers:", len(hv))
print("High-value churners:", hv["Exited"].sum())
print("High-Value Churn Ratio (%):", hv["Exited"].mean() * 100)
```

```python
total_balance = df_clean["Balance"].sum()
churned_balance = df_clean.loc[df_clean["Exited"] == 1, "Balance"].sum()
print("Balance held by churned customers:", churned_balance)
print("Share of all balances (%):", churned_balance / total_balance * 100)
```

```python
engagement = df_clean.groupby("IsActiveMember")["Exited"].mean() * 100
print(engagement)
print("Inactive vs active ratio:", engagement[0] / engagement[1])
```

```python
df_clean["SalaryBand"] = pd.qcut(
    df_clean["EstimatedSalary"], 4,
    labels=["Q1 (lowest)", "Q2", "Q3", "Q4 (highest)"]
)
print(df_clean.groupby("SalaryBand", observed=True)["Exited"].mean() * 100)
```
